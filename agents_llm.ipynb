```python

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
from langchain_classic.agents import create_react_agent, AgentExecutor
from langchain_community.llms import HuggingFacePipeline
from langchain_core.prompts import PromptTemplate
from langchain_core.tools import Tool

# 1. Load a small open-source instruction model
model_id = "Qwen/Qwen2.5-0.5B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    torch_dtype="auto"
)

pipe = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=128,
    do_sample=False,
    return_full_text=False
)

llm = HuggingFacePipeline(pipeline=pipe)

# 2. Define a tool
def multiply_numbers(text):
    a, b = map(float, text.split(","))
    result = a * b
    return str(int(result)) if result.is_integer() else str(result)

tools = [
    Tool(
        name="Multiply",
        func=multiply_numbers,
        description="Multiply two numbers. Input must look like 25,16."
    )
]

# 3. Define the ReAct prompt locally
prompt = PromptTemplate.from_template("""You are a ReAct agent.
Use a tool when needed.

Available tools:
{tools}

Tool names: {tool_names}

Follow this format exactly:
Question: the question
Thought: briefly explain the next step
Action: one tool name from [{tool_names}]
Action Input: valid tool input
Observation: tool result
Thought: I know the answer
Final Answer: the answer

Question: {input}
{agent_scratchpad}""")

# 4. Create the ReAct agent
agent = create_react_agent(llm, tools, prompt) #one of the agent of all agent

executor = AgentExecutor(
    agent=agent,
    tools=tools,
    verbose=True,
    handle_parsing_errors=True,
    max_iterations=4
)

# 5. Run the agent
result = executor.invoke({
    "input": "What is 25 multiplied by 16?"
})

print("Final answer:", result["output"])
```

```python
import ast
import operator
import requests
import ollama
import subprocess
import time

MODEL = "llama3.1"
MAX_STEPS = 6


# Start Ollama background server if not already running
try:
    # Try to ping the server
    requests.get("http://localhost:11434")
except requests.exceptions.ConnectionError:
    print("Ollama server is not running. Starting it in the background...")
    subprocess.Popen(["ollama", "serve"], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
    # Wait for the server to spin up
    time.sleep(5)

# Ensure the model is downloaded
try:
    local_models = ollama.list()
    model_names = [m.model for m in local_models.models]
    if not any(MODEL in name for name in model_names):
        print(f"Pulling '{MODEL}' model (this might take a few minutes)... ")
        ollama.pull(MODEL)
        print("Model pulled successfully!")
except Exception as e:
    print(f"Pulling model via API failed ({e}). Attempting terminal pull...")
    subprocess.run(["ollama", "pull", MODEL])


# TOOL 1: Search Wikipedia
def search_wikipedia(query: str) -> str:
    """Search Wikipedia and return relevant information."""

    response = requests.get(
        "https://en.wikipedia.org/w/api.php",
        params={
            "action": "query",
            "list": "search",
            "srsearch": query,
            "format": "json",
        },
        headers={"User-Agent": "LocalReActAgent/1.0"},
        timeout=15,
    )
    response.raise_for_status()
    results = response.json()["query"]["search"]

    if not results:
        return "No Wikipedia results found."

    output = []

    for result in results[:3]:
        output.append(
            f"Title: {result['title']}\n"
            f"Snippet: {result['snippet']}"
        )

    return "\n\n".join(output)


# TOOL 2: Safe arithmetic calculator
def calculate(expression: str) -> str:
    """Calculate basic arithmetic expressions."""

    allowed_operators = {
        ast.Add: operator.add,
        ast.Sub: operator.sub,
        ast.Mult: operator.mul,
        ast.Div: operator.truediv,
        ast.Pow: operator.pow,
        ast.USub: operator.neg,
        ast.UAdd: operator.pos,
    }

    def evaluate(node):
        if isinstance(node, ast.Constant) and type(node.value) in (int, float):
            return node.value

        if isinstance(node, ast.BinOp):
            operation = allowed_operators.get(type(node.op))

            if operation is None:
                raise ValueError("Unsupported operator")

            left = evaluate(node.left)
            right = evaluate(node.right)

            # Prevent excessively large calculations
            if abs(left) > 1e100 or abs(right) > 1e100:
                raise ValueError("Number is too large")

            if isinstance(node.op, ast.Pow) and abs(right) > 10:
                raise ValueError("Exponent is too large")

            return operation(left, right)

        if isinstance(node, ast.UnaryOp):
            operation = allowed_operators.get(type(node.op))

            if operation is None:
                raise ValueError("Unsupported operator")

            return operation(evaluate(node.operand))

        raise ValueError("Invalid arithmetic expression")

    try:
        tree = ast.parse(expression, mode="eval")
        result = evaluate(tree.body)
        return str(result)
    except Exception as error:
        return f"Calculation error: {error}"


# Available tools
TOOLS = [search_wikipedia, calculate]

AVAILABLE_FUNCTIONS = {
    "search_wikipedia": search_wikipedia,
    "calculate": calculate,
}


def run_react_agent(question: str):
    messages = [
        {
            "role": "system",
            "content": (
                "You are a ReAct-style assistant. "
                "Use tools when external information or calculation "
                "is needed. You may call multiple tools in sequence. "
                "After receiving tool results, answer accurately. "
                "Do not invent tool observations."
            ),
        },
        {"role": "user", "content": question},
    ]

    print(f"\nUSER: {question}")

    for step in range(1, MAX_STEPS + 1):
        response = ollama.chat(
            model=MODEL,
            messages=messages,
            tools=TOOLS,
        )

        assistant_message = response.message
        messages.append(assistant_message)

        tool_calls = assistant_message.tool_calls or []

        # No more tool calls: the model has produced its answer.
        if not tool_calls:
            print("\nFINAL ANSWER:")
            print(assistant_message.content)
            return assistant_message.content

        for tool_call in tool_calls:
            tool_name = tool_call.function.name
            arguments = tool_call.function.arguments

            # Print a short decision summary, not private reasoning.
            print(f"\n--- STEP {step} ---")
            print(
                "THOUGHT: The model decided to use "
                f"{tool_name} to obtain the required information."
            )

            print("ACTION:")
            print(f"Tool: {tool_name}")
            print(f"Input: {arguments}")

            function = AVAILABLE_FUNCTIONS.get(tool_name)

            if function is None:
                observation = f"Unknown tool: {tool_name}"
            else:
                try:
                    observation = function(**arguments)
                except Exception as error:
                    observation = f"Tool error: {error}"

            print("\nOBSERVATION:")
            print(observation)

            # Feed the real tool result back to the model.
            messages.append({
                "role": "tool",
                "tool_name": tool_name,
                "content": str(observation),
            })

    print("\nStopped: maximum tool steps reached.")
    return None


if __name__ == "__main__":
    question = (
        "Find the capital of France using Wikipedia, "
        "then calculate 125 * 8."
    )

    run_react_agent(question)
```
